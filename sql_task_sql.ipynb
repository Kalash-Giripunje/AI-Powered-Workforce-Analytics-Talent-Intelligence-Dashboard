```python
# ==========================================================
#           EMPLOYEE MANAGEMENT SYSTEM
# ==========================================================
# SECTION 2 : TABLE BASICS
# ==========================================================

# ----------------------------------------------------------
# Objective:
# Learn how to create, describe, rename, drop,
# and truncate (delete all records) a table.
# ----------------------------------------------------------

import sqlite3

# Connect to the database
connection = sqlite3.connect("employee_management_system.db")

# Create a cursor object
cursor = connection.cursor()

print("=" * 60)
print("SECTION 2 : TABLE BASICS")
print("=" * 60)

# ==========================================================
# 1. CREATE TABLE
# ==========================================================
# A table stores data in rows and columns.

cursor.execute("""
CREATE TABLE IF NOT EXISTS Employees(
    employee_id INTEGER PRIMARY KEY,
    employee_name TEXT NOT NULL,
    department TEXT,
    designation TEXT,
    salary REAL
)
""")

connection.commit()

print("\nEmployees table created successfully.")

# Expected Output:
# Employees table created successfully.


# ==========================================================
# 2. DESCRIBE TABLE
# ==========================================================
# SQLite uses PRAGMA table_info() to display
# table structure.

print("\n")
print("=" * 60)
print("DESCRIBE TABLE")
print("=" * 60)

cursor.execute("PRAGMA table_info(Employees)")

columns = cursor.fetchall()

for column in columns:
    print(column)

# Expected Output:
# (0, 'employee_id', 'INTEGER', ...)
# (1, 'employee_name', 'TEXT', ...)
# (2, 'department', 'TEXT', ...)
# (3, 'designation', 'TEXT', ...)
# (4, 'salary', 'REAL', ...)


# ==========================================================
# 3. RENAME TABLE
# ==========================================================
# Rename Employees table to EmployeeDetails.

cursor.execute("""
ALTER TABLE Employees
RENAME TO EmployeeDetails;
""")

connection.commit()

print("\nTable renamed successfully.")

# Expected Output:
# Table renamed successfully.


# ==========================================================
# 4. RENAME BACK
# ==========================================================
# Rename it back so future sections continue working.

cursor.execute("""
ALTER TABLE EmployeeDetails
RENAME TO Employees;
""")

connection.commit()

print("Table renamed back to Employees.")

# Expected Output:
# Table renamed back to Employees.


# ==========================================================
# 5. DROP TABLE
# ==========================================================
# DROP TABLE permanently deletes the table.

cursor.execute("""
DROP TABLE IF EXISTS TemporaryEmployees;
""")

connection.commit()

print("\nDROP TABLE example executed.")

# Expected Output:
# DROP TABLE example executed.


# ==========================================================
# 6. TRUNCATE TABLE (SQLite Alternative)
# ==========================================================
# SQLite does not support TRUNCATE TABLE.
# DELETE removes all records from the table.

cursor.execute("DELETE FROM Employees")

connection.commit()

print("All records deleted from Employees table.")

# Expected Output:
# All records deleted from Employees table.


# Close the connection
connection.close()
```

```python
# ==========================================================
#           EMPLOYEE MANAGEMENT SYSTEM
# ==========================================================
# SECTION 4 : SELECT STATEMENT
# ==========================================================

# ----------------------------------------------------------
# Objective:
# Learn how to retrieve data from a table using
# the SELECT statement.
#
# Topics Covered:
# 1. SELECT *
# 2. SELECT Specific Columns
# ----------------------------------------------------------

import sqlite3

# Connect to the database
connection = sqlite3.connect("employee_management_system.db")

# Create a cursor object
cursor = connection.cursor()

print("=" * 60)
print("SECTION 4 : SELECT STATEMENT")
print("=" * 60)

# ==========================================================
# 1. SELECT *
# ==========================================================
# SELECT * retrieves all columns from the Employees table.

print("\n")
print("=" * 60)
print("1. SELECT *")
print("=" * 60)

cursor.execute("""
SELECT *
FROM Employees;
""")

employees = cursor.fetchall()

for employee in employees:
    print(employee)

# Expected Output:
#
# (101, 'Rahul Sharma', 'Software Development', 65000.0)
# (102, 'Priya Verma', 'Human Resources', 55000.0)
# (103, 'Amit Patel', 'Finance', 70000.0)
# (104, 'Sneha Joshi', 'Marketing', 60000.0)
# (105, 'Rohit Kumar', 'Testing', None)


# ==========================================================
# 2. SELECT SPECIFIC COLUMNS
# ==========================================================
# Retrieve only selected columns instead of
# displaying the entire table.

print("\n")
print("=" * 60)
print("2. SELECT SPECIFIC COLUMNS")
print("=" * 60)

cursor.execute("""
SELECT
    employee_name,
    department,
    salary
FROM Employees;
""")

employees = cursor.fetchall()

for employee in employees:
    print(employee)

# Expected Output:
#
# ('Rahul Sharma', 'Software Development', 65000.0)
# ('Priya Verma', 'Human Resources', 55000.0)
# ('Amit Patel', 'Finance', 70000.0)
# ('Sneha Joshi', 'Marketing', 60000.0)
# ('Rohit Kumar', 'Testing', None)


# ==========================================================
# Real-World Example
# ==========================================================
# Display employee details in a readable format.

print("\n")
print("=" * 60)
print("EMPLOYEE DIRECTORY")
print("=" * 60)

cursor.execute("""
SELECT
    employee_name,
    department,
    salary
FROM Employees;
""")

employee_list = cursor.fetchall()

for employee in employee_list:

    print(f"Employee Name : {employee[0]}")
    print(f"Department    : {employee[1]}")
    print(f"Salary        : ₹{employee[2]}")
    print("-" * 40)

# Expected Output:
#
# Employee Name : Rahul Sharma
# Department    : Software Development
# Salary        : ₹65000.0
# ----------------------------------------
#
# Employee Name : Priya Verma
# Department    : Human Resources
# Salary        : ₹55000.0
# ----------------------------------------
#
# Employee Name : Amit Patel
# Department    : Finance
# Salary        : ₹70000.0
# ----------------------------------------
#
# Employee Name : Sneha Joshi
# Department    : Marketing
# Salary        : ₹60000.0
# ----------------------------------------
#
# Employee Name : Rohit Kumar
# Department    : Testing
# Salary        : ₹None
# ----------------------------------------

# Close the database connection
connection.close()
```
