```python
# ==========================================================
#           EMPLOYEE MANAGEMENT SYSTEM
# ==========================================================
# SECTION 1 : VARIABLES AND DATA TYPES
# ==========================================================

# ----------------------------------------------------------
# Objective:
# Learn how different Python data types are used in an
# Employee Management System.
# ----------------------------------------------------------

print("=" * 60)
print("SECTION 1 : VARIABLES AND DATA TYPES")
print("=" * 60)

# ==========================================================
# 1. INTEGER (int)
# ==========================================================
# Integer values store whole numbers.

employee_id = 1001
employee_age = 28
employee_experience = 5

print("\nInteger Variables")
print("-----------------")
print("Employee ID:", employee_id)
print("Employee Age:", employee_age)
print("Experience (Years):", employee_experience)

print("Type of employee_id:", type(employee_id))
print("Type of employee_age:", type(employee_age))
print("Type of employee_experience:", type(employee_experience))

# Expected Output:
# Employee ID: 1001
# Employee Age: 28
# Experience (Years): 5
# Type of employee_id: <class 'int'>
# Type of employee_age: <class 'int'>
# Type of employee_experience: <class 'int'>


# ==========================================================
# 2. FLOAT (float)
# ==========================================================
# Float values store decimal numbers.

employee_salary = 65000.75
employee_bonus = 8500.50

print("\nFloat Variables")
print("----------------")
print("Salary:", employee_salary)
print("Bonus:", employee_bonus)

print("Type of employee_salary:", type(employee_salary))
print("Type of employee_bonus:", type(employee_bonus))

# Expected Output:
# Salary: 65000.75
# Bonus: 8500.5
# Type of employee_salary: <class 'float'>
# Type of employee_bonus: <class 'float'>


# ==========================================================
# 3. STRING (str)
# ==========================================================
# Strings store text.

employee_name = "Rahul Sharma"
employee_department = "Software Development"
employee_designation = "Software Engineer"
employee_email = "rahul.sharma@company.com"

print("\nString Variables")
print("----------------")
print("Employee Name:", employee_name)
print("Department:", employee_department)
print("Designation:", employee_designation)
print("Email:", employee_email)

print("Type of employee_name:", type(employee_name))
print("Type of employee_department:", type(employee_department))
print("Type of employee_designation:", type(employee_designation))
print("Type of employee_email:", type(employee_email))

# Expected Output:
# Employee Name: Rahul Sharma
# Department: Software Development
# Designation: Software Engineer
# Email: rahul.sharma@company.com
# Type of employee_name: <class 'str'>


# ==========================================================
# 4. BOOLEAN (bool)
# ==========================================================
# Boolean values store True or False.

employee_is_active = True
employee_is_manager = False

print("\nBoolean Variables")
print("-----------------")
print("Employee Active:", employee_is_active)
print("Employee Manager:", employee_is_manager)

print("Type of employee_is_active:", type(employee_is_active))
print("Type of employee_is_manager:", type(employee_is_manager))

# Expected Output:
# Employee Active: True
# Employee Manager: False
# Type of employee_is_active: <class 'bool'>
# Type of employee_is_manager: <class 'bool'>


# ==========================================================
# 5. LIST (list)
# ==========================================================
# Lists store multiple values in a single variable.

employee_skills = [
    "Python",
    "SQL",
    "AWS",
    "Git"
]

print("\nList Variable")
print("-------------")
print("Employee Skills:", employee_skills)

print("First Skill:", employee_skills[0])
print("Second Skill:", employee_skills[1])
print("Total Skills:", len(employee_skills))

print("Type of employee_skills:", type(employee_skills))

# Expected Output:
# Employee Skills: ['Python', 'SQL', 'AWS', 'Git']
# First Skill: Python
# Second Skill: SQL
# Total Skills: 4
# Type of employee_skills: <class 'list'>


# ==========================================================
# Display Complete Employee Information
# ==========================================================

print("\nComplete Employee Details")
print("-" * 30)

print("Employee ID        :", employee_id)
print("Employee Name      :", employee_name)
print("Employee Age       :", employee_age)
print("Department         :", employee_department)
print("Designation        :", employee_designation)
print("Salary             :", employee_salary)
print("Bonus              :", employee_bonus)
print("Experience         :", employee_experience)
print("Active Employee    :", employee_is_active)
print("Skills             :", employee_skills)

# Expected Output:
# Employee ID        : 1001
# Employee Name      : Rahul Sharma
# Employee Age       : 28
# Department         : Software Development
# Designation        : Software Engineer
# Salary             : 65000.75
# Bonus              : 8500.5
# Experience         : 5
# Active Employee    : True
# Skills             : ['Python', 'SQL', 'AWS', 'Git']


# ==========================================================
# Summary
# ==========================================================

print("\nSummary")
print("-" * 30)

print(f"""
Integer  : Employee ID, Age, Experience
Float    : Salary, Bonus
String   : Name, Department, Designation, Email
Boolean  : Active Status, Manager Status
List     : Employee Skills
""")

# Expected Output:
#
# Summary
# ------------------------------
# Integer  : Employee ID, Age, Experience
# Float    : Salary, Bonus
# String   : Name, Department, Designation, Email
# Boolean  : Active Status, Manager Status
# List     : Employee Skills
```

```python
# ==========================================================
#           EMPLOYEE MANAGEMENT SYSTEM
# ==========================================================
# SECTION 3 : OPERATORS
# ==========================================================

# ----------------------------------------------------------
# Objective:
# Learn how different types of operators are used in an
# Employee Management System.
#
# Types of Operators Covered:
# 1. Arithmetic Operators
# 2. Comparison Operators
# 3. Logical Operators
# 4. Assignment Operators
# 5. Membership Operators
# 6. Identity Operators
# ----------------------------------------------------------

print("=" * 60)
print("SECTION 3 : OPERATORS")
print("=" * 60)

# ==========================================================
# Employee Data
# ==========================================================

employee_name = "Rahul Sharma"
employee_salary = 65000.00
employee_bonus = 5000.00
employee_experience = 5
employee_department = "Software Development"
employee_is_active = True
employee_skills = ["Python", "SQL", "AWS", "Git"]

print("\nEmployee Information")
print("-" * 30)
print(f"Employee Name : {employee_name}")
print(f"Salary        : ₹{employee_salary}")
print(f"Bonus         : ₹{employee_bonus}")
print(f"Experience    : {employee_experience} Years")
print(f"Department    : {employee_department}")
print(f"Skills        : {employee_skills}")

# ==========================================================
# 1. Arithmetic Operators
# ==========================================================
# Arithmetic operators are used to perform mathematical
# calculations.

print("\n")
print("=" * 60)
print("1. Arithmetic Operators")
print("=" * 60)

annual_salary = employee_salary * 12
total_income = employee_salary + employee_bonus
salary_after_deduction = employee_salary - 2000
monthly_salary = annual_salary / 12
experience_double = employee_experience * 2
remaining_salary = employee_salary % 7000

print(f"Annual Salary           : ₹{annual_salary}")
print(f"Total Income            : ₹{total_income}")
print(f"Salary After Deduction  : ₹{salary_after_deduction}")
print(f"Monthly Salary          : ₹{monthly_salary}")
print(f"Double Experience       : {experience_double}")
print(f"Salary Modulus 7000     : {remaining_salary}")

# Expected Output:
# Annual Salary           : ₹780000.0
# Total Income            : ₹70000.0
# Salary After Deduction  : ₹63000.0
# Monthly Salary          : ₹65000.0
# Double Experience       : 10
# Salary Modulus 7000     : 2000.0


# ==========================================================
# 2. Comparison Operators
# ==========================================================
# Comparison operators compare two values and return either
# True or False.

print("\n")
print("=" * 60)
print("2. Comparison Operators")
print("=" * 60)

print(f"Salary > 50000           : {employee_salary > 50000}")
print(f"Salary < 50000           : {employee_salary < 50000}")
print(f"Experience >= 5          : {employee_experience >= 5}")
print(f"Experience <= 3          : {employee_experience <= 3}")
print(f"Department == HR         : {employee_department == 'HR'}")
print(f"Department != HR         : {employee_department != 'HR'}")

# Expected Output:
# Salary > 50000           : True
# Salary < 50000           : False
# Experience >= 5          : True
# Experience <= 3          : False
# Department == HR         : False
# Department != HR         : True


# ==========================================================
# 3. Logical Operators
# ==========================================================
# Logical operators combine multiple conditions.

print("\n")
print("=" * 60)
print("3. Logical Operators")
print("=" * 60)

eligible_for_bonus = employee_salary > 50000 and employee_experience >= 5
eligible_for_training = employee_department == "HR" or employee_department == "Software Development"
employee_not_active = not employee_is_active

print(f"Eligible for Bonus       : {eligible_for_bonus}")
print(f"Eligible for Training    : {eligible_for_training}")
print(f"Employee Not Active      : {employee_not_active}")

# Expected Output:
# Eligible for Bonus       : True
# Eligible for Training    : True
# Employee Not Active      : False


# ==========================================================
# 4. Assignment Operators
# ==========================================================
# Assignment operators update variable values.

print("\n")
print("=" * 60)
print("4. Assignment Operators")
print("=" * 60)

salary = 50000

print(f"Original Salary : ₹{salary}")

salary += 5000
print(f"After += 5000   : ₹{salary}")

salary -= 2000
print(f"After -= 2000   : ₹{salary}")

salary *= 2
print(f"After *= 2      : ₹{salary}")

salary /= 2
print(f"After /= 2      : ₹{salary}")

salary //= 1000
print(f"After //=1000   : {salary}")

# Expected Output:
# Original Salary : ₹50000
# After += 5000   : ₹55000
# After -= 2000   : ₹53000
# After *= 2      : ₹106000
# After /= 2      : ₹53000.0
# After //=1000   : 53.0


# ==========================================================
# 5. Membership Operators
# ==========================================================
# Membership operators check whether a value exists inside
# a sequence such as a list.

print("\n")
print("=" * 60)
print("5. Membership Operators")
print("=" * 60)

print(f"'Python' in Skills       : {'Python' in employee_skills}")
print(f"'Java' in Skills         : {'Java' in employee_skills}")
print(f"'Java' not in Skills     : {'Java' not in employee_skills}")

# Expected Output:
# 'Python' in Skills       : True
# 'Java' in Skills         : False
# 'Java' not in Skills     : True


# ==========================================================
# 6. Identity Operators
# ==========================================================
# Identity operators compare whether two variables refer
# to the same object in memory.

print("\n")
print("=" * 60)
print("6. Identity Operators")
print("=" * 60)

skills_list = employee_skills
new_skills_list = ["Python", "SQL", "AWS", "Git"]

print(f"skills_list is employee_skills      : {skills_list is employee_skills}")
print(f"skills_list == employee_skills      : {skills_list == employee_skills}")

print(f"new_skills_list is employee_skills  : {new_skills_list is employee_skills}")
print(f"new_skills_list == employee_skills  : {new_skills_list == employee_skills}")

# Expected Output:
# skills_list is employee_skills      : True
# skills_list == employee_skills      : True
# new_skills_list is employee_skills  : False
# new_skills_list == employee_skills  : True


# ==========================================================
# Summary
# ==========================================================

print("\n")
print("=" * 60)
print("Summary")
print("=" * 60)

print("✔ Arithmetic Operators  : Perform mathematical calculations")
print("✔ Comparison Operators  : Compare two values")
print("✔ Logical Operators     : Combine multiple conditions")
print("✔ Assignment Operators  : Update variable values")
print("✔ Membership Operators  : Check whether a value exists")
print("✔ Identity Operators    : Check whether two variables refer to the same object")

# Expected Output:
#
# ==========================================================
# Summary
# ==========================================================
# ✔ Arithmetic Operators  : Perform mathematical calculations
# ✔ Comparison Operators  : Compare two values
# ✔ Logical Operators     : Combine multiple conditions
# ✔ Assignment Operators  : Update variable values
# ✔ Membership Operators  : Check whether a value exists
# ✔ Identity Operators    : Check whether two variables refer to the same object
```

```python
# ==========================================================
#           EMPLOYEE MANAGEMENT SYSTEM
# ==========================================================
# SECTION 4 : CONDITIONAL STATEMENTS
# ==========================================================

# ----------------------------------------------------------
# Objective:
# Learn how conditional statements help make decisions in an
# Employee Management System.
#
# Topics Covered:
# 1. if Statement
# 2. if-else Statement
# 3. if-elif-else Statement
# 4. Nested if Statement
# ----------------------------------------------------------

print("=" * 60)
print("SECTION 4 : CONDITIONAL STATEMENTS")
print("=" * 60)

# ==========================================================
# Employee Data
# ==========================================================

employee_id = 1001
employee_name = "Rahul Sharma"
employee_age = 28
employee_department = "Software Development"
employee_designation = "Software Engineer"
employee_salary = 65000
employee_experience = 5
employee_is_active = True
employee_performance_rating = 4.5

print("\nEmployee Details")
print("-" * 30)
print(f"Employee ID          : {employee_id}")
print(f"Employee Name        : {employee_name}")
print(f"Department           : {employee_department}")
print(f"Designation          : {employee_designation}")
print(f"Salary               : ₹{employee_salary}")
print(f"Experience           : {employee_experience} Years")
print(f"Performance Rating   : {employee_performance_rating}")
print(f"Active Employee      : {employee_is_active}")

# ==========================================================
# 1. IF Statement
# ==========================================================
# Executes a block only if the condition is True.

print("\n")
print("=" * 60)
print("1. IF Statement")
print("=" * 60)

if employee_salary >= 60000:
    print(f"{employee_name} earns more than ₹60,000 per month.")

# Expected Output:
# Rahul Sharma earns more than ₹60,000 per month.


# ==========================================================
# 2. IF-ELSE Statement
# ==========================================================
# Executes one block if the condition is True,
# otherwise executes the else block.

print("\n")
print("=" * 60)
print("2. IF-ELSE Statement")
print("=" * 60)

if employee_is_active:
    print(f"{employee_name} is currently an active employee.")
else:
    print(f"{employee_name} is no longer working in the company.")

# Expected Output:
# Rahul Sharma is currently an active employee.


# ==========================================================
# 3. IF-ELIF-ELSE Statement
# ==========================================================
# Used when multiple conditions need to be checked.

print("\n")
print("=" * 60)
print("3. IF-ELIF-ELSE Statement")
print("=" * 60)

if employee_performance_rating >= 4.5:
    performance = "Excellent"
elif employee_performance_rating >= 3.5:
    performance = "Good"
elif employee_performance_rating >= 2.5:
    performance = "Average"
else:
    performance = "Needs Improvement"

print(f"Performance Rating : {employee_performance_rating}")
print(f"Performance Level  : {performance}")

# Expected Output:
# Performance Rating : 4.5
# Performance Level  : Excellent


# ==========================================================
# 4. Nested IF Statement
# ==========================================================
# An if statement inside another if statement.

print("\n")
print("=" * 60)
print("4. Nested IF Statement")
print("=" * 60)

if employee_is_active:

    print(f"{employee_name} is an active employee.")

    if employee_experience >= 5:

        print("Employee is eligible for promotion review.")

        if employee_performance_rating >= 4.5:
            print("Employee is eligible for promotion and bonus.")
        else:
            print("Employee is eligible for promotion only.")

    else:
        print("Employee needs more experience before promotion.")

else:
    print("Inactive employees are not eligible for promotion.")

# Expected Output:
# Rahul Sharma is an active employee.
# Employee is eligible for promotion review.
# Employee is eligible for promotion and bonus.


# ==========================================================
# Real-World Example
# ==========================================================
# HR Bonus Eligibility Check

print("\n")
print("=" * 60)
print("HR Bonus Eligibility")
print("=" * 60)

if employee_is_active and employee_salary >= 50000:

    if employee_performance_rating >= 4.5:
        bonus = 10000

    elif employee_performance_rating >= 3.5:
        bonus = 5000

    else:
        bonus = 2000

    print(f"{employee_name} will receive a bonus of ₹{bonus}.")

else:
    print(f"{employee_name} is not eligible for a bonus.")

# Expected Output:
# Rahul Sharma will receive a bonus of ₹10000.


# ==========================================================
# Summary
# ==========================================================

print("\n")
print("=" * 60)
print("Summary")
print("=" * 60)

print("✔ if                : Executes when a condition is True.")
print("✔ if-else           : Chooses between two alternatives.")
print("✔ if-elif-else      : Handles multiple conditions.")
print("✔ Nested if         : Places one if statement inside another.")
print("✔ Conditional statements are commonly used for")
print("  salary calculation, bonus eligibility,")
print("  promotions, attendance, and performance evaluation.")

# Expected Output:
#
# ==========================================================
# Summary
# ==========================================================
# ✔ if                : Executes when a condition is True.
# ✔ if-else           : Chooses between two alternatives.
# ✔ if-elif-else      : Handles multiple conditions.
# ✔ Nested if         : Places one if statement inside another.
# ✔ Conditional statements are commonly used for
#   salary calculation, bonus eligibility,
#   promotions, attendance, and performance evaluation.
```

```python
# ==========================================================
#           EMPLOYEE MANAGEMENT SYSTEM
# ==========================================================
# SECTION 5 : LOOPS
# ==========================================================

# ----------------------------------------------------------
# Objective:
# Learn how loops are used to repeatedly execute a block of
# code in an Employee Management System.
#
# Topics Covered:
# 1. for Loop
# 2. while Loop
# 3. break Statement
# 4. continue Statement
# 5. pass Statement
# ----------------------------------------------------------

print("=" * 60)
print("SECTION 5 : LOOPS")
print("=" * 60)

# ==========================================================
# Employee Data
# ==========================================================

employees = [
    "Rahul Sharma",
    "Priya Verma",
    "Amit Patel",
    "Sneha Joshi",
    "Rohit Kumar"
]

departments = [
    "Software Development",
    "Human Resources",
    "Finance",
    "Marketing",
    "Testing"
]

employee_salaries = [65000, 55000, 70000, 48000, 60000]

# ==========================================================
# 1. FOR LOOP
# ==========================================================
# A for loop is used to iterate over a sequence such as a
# list, tuple, string, or dictionary.

print("\n")
print("=" * 60)
print("1. FOR LOOP")
print("=" * 60)

print("Employee Names")

for employee in employees:
    print(employee)

# Expected Output:
# Employee Names
# Rahul Sharma
# Priya Verma
# Amit Patel
# Sneha Joshi
# Rohit Kumar


print("\nEmployee Name and Department")

for employee, department in zip(employees, departments):
    print(f"{employee} works in {department}")

# Expected Output:
# Rahul Sharma works in Software Development
# Priya Verma works in Human Resources
# Amit Patel works in Finance
# Sneha Joshi works in Marketing
# Rohit Kumar works in Testing


# ==========================================================
# 2. WHILE LOOP
# ==========================================================
# A while loop executes as long as the condition remains True.

print("\n")
print("=" * 60)
print("2. WHILE LOOP")
print("=" * 60)

employee_number = 1

while employee_number <= 5:
    print(f"Processing Employee {employee_number}")
    employee_number += 1

# Expected Output:
# Processing Employee 1
# Processing Employee 2
# Processing Employee 3
# Processing Employee 4
# Processing Employee 5


# ==========================================================
# 3. BREAK STATEMENT
# ==========================================================
# The break statement immediately terminates the loop.

print("\n")
print("=" * 60)
print("3. BREAK STATEMENT")
print("=" * 60)

for employee in employees:

    if employee == "Amit Patel":
        print("Target employee found.")
        break

    print(employee)

# Expected Output:
# Rahul Sharma
# Priya Verma
# Target employee found.


# ==========================================================
# 4. CONTINUE STATEMENT
# ==========================================================
# The continue statement skips the current iteration and
# moves to the next iteration.

print("\n")
print("=" * 60)
print("4. CONTINUE STATEMENT")
print("=" * 60)

for salary in employee_salaries:

    if salary < 60000:
        continue

    print(f"Eligible Salary : ₹{salary}")

# Expected Output:
# Eligible Salary : ₹65000
# Eligible Salary : ₹70000
# Eligible Salary : ₹60000


# ==========================================================
# 5. PASS STATEMENT
# ==========================================================
# The pass statement does nothing. It is used as a placeholder
# when code will be added later.

print("\n")
print("=" * 60)
print("5. PASS STATEMENT")
print("=" * 60)

for employee in employees:

    if employee == "Sneha Joshi":
        pass
    else:
        print(employee)

print("\nLoop completed successfully.")

# Expected Output:
# Rahul Sharma
# Priya Verma
# Amit Patel
# Rohit Kumar
#
# Loop completed successfully.


# ==========================================================
# Real-World Example
# ==========================================================
# Display employees who are eligible for a salary increment.
# Employees earning ₹60,000 or more are eligible.

print("\n")
print("=" * 60)
print("Employees Eligible for Salary Increment")
print("=" * 60)

for employee, salary in zip(employees, employee_salaries):

    if salary >= 60000:
        print(f"{employee} is eligible (Salary: ₹{salary})")

# Expected Output:
# Rahul Sharma is eligible (Salary: ₹65000)
# Amit Patel is eligible (Salary: ₹70000)
# Rohit Kumar is eligible (Salary: ₹60000)
```

```python
# ==========================================================
#           EMPLOYEE MANAGEMENT SYSTEM
# ==========================================================
# SECTION 6 : FUNCTIONS
# ==========================================================

# ----------------------------------------------------------
# Objective:
# Learn how functions help organize and reuse code in an
# Employee Management System.
#
# Topics Covered:
# 1. Function with Parameters
# 2. Return Values
# 3. Default Arguments
# 4. Keyword Arguments
# ----------------------------------------------------------

print("=" * 60)
print("SECTION 6 : FUNCTIONS")
print("=" * 60)

# ==========================================================
# Employee Data
# ==========================================================

employee_id = 1001
employee_name = "Rahul Sharma"
employee_department = "Software Development"
employee_salary = 65000
employee_experience = 5

# ==========================================================
# 1. FUNCTION WITH PARAMETERS
# ==========================================================
# A function can accept values called parameters.
# These parameters allow the function to work with
# different employee information.

print("\n")
print("=" * 60)
print("1. FUNCTION WITH PARAMETERS")
print("=" * 60)


def display_employee(employee_id, employee_name, department):
    print(f"Employee ID   : {employee_id}")
    print(f"Employee Name : {employee_name}")
    print(f"Department    : {department}")


display_employee(employee_id, employee_name, employee_department)

# Expected Output:
# Employee ID   : 1001
# Employee Name : Rahul Sharma
# Department    : Software Development


# ==========================================================
# 2. RETURN VALUES
# ==========================================================
# A function can return a value back to the caller.

print("\n")
print("=" * 60)
print("2. RETURN VALUES")
print("=" * 60)


def calculate_annual_salary(monthly_salary):
    annual_salary = monthly_salary * 12
    return annual_salary


annual_salary = calculate_annual_salary(employee_salary)

print(f"Monthly Salary : ₹{employee_salary}")
print(f"Annual Salary  : ₹{annual_salary}")

# Expected Output:
# Monthly Salary : ₹65000
# Annual Salary  : ₹780000


# ==========================================================
# 3. DEFAULT ARGUMENTS
# ==========================================================
# Default arguments provide a value automatically if
# no value is supplied while calling the function.

print("\n")
print("=" * 60)
print("3. DEFAULT ARGUMENTS")
print("=" * 60)


def generate_employee_email(name, domain="company.com"):
    email = name.lower().replace(" ", ".") + "@" + domain
    return email


print("Using Default Domain")
print(generate_employee_email(employee_name))

print("\nUsing Custom Domain")
print(generate_employee_email(employee_name, "techcorp.com"))

# Expected Output:
# Using Default Domain
# rahul.sharma@company.com
#
# Using Custom Domain
# rahul.sharma@techcorp.com


# ==========================================================
# 4. KEYWORD ARGUMENTS
# ==========================================================
# Keyword arguments allow passing values using the
# parameter names. The order of arguments does not matter.

print("\n")
print("=" * 60)
print("4. KEYWORD ARGUMENTS")
print("=" * 60)


def employee_profile(employee_id,
                     employee_name,
                     department,
                     salary,
                     experience):

    print(f"Employee ID : {employee_id}")
    print(f"Employee    : {employee_name}")
    print(f"Department  : {department}")
    print(f"Salary      : ₹{salary}")
    print(f"Experience  : {experience} Years")


employee_profile(
    salary=65000,
    experience=5,
    employee_name="Rahul Sharma",
    employee_id=1001,
    department="Software Development"
)

# Expected Output:
# Employee ID : 1001
# Employee    : Rahul Sharma
# Department  : Software Development
# Salary      : ₹65000
# Experience  : 5 Years


# ==========================================================
# Real-World Example
# ==========================================================
# Calculate the salary after adding a performance bonus.

print("\n")
print("=" * 60)
print("Employee Salary Calculator")
print("=" * 60)


def calculate_total_salary(monthly_salary, bonus=5000):
    total_salary = monthly_salary + bonus
    return total_salary


total_salary = calculate_total_salary(employee_salary)

print(f"Employee Name : {employee_name}")
print(f"Base Salary   : ₹{employee_salary}")
print(f"Bonus         : ₹5000")
print(f"Total Salary  : ₹{total_salary}")

# Expected Output:
# Employee Name : Rahul Sharma
# Base Salary   : ₹65000
# Bonus         : ₹5000
# Total Salary  : ₹70000


# ==========================================================
# Another Example
# ==========================================================
# Calculate the years remaining until retirement.
# Assume retirement age is 60 years.

print("\n")
print("=" * 60)
print("Retirement Calculator")
print("=" * 60)


def years_to_retirement(current_age, retirement_age=60):
    return retirement_age - current_age


remaining_years = years_to_retirement(28)

print(f"Employee Name           : {employee_name}")
print(f"Current Age             : 28")
print(f"Years Until Retirement  : {remaining_years}")

# Expected Output:
# Employee Name           : Rahul Sharma
# Current Age             : 28
# Years Until Retirement  : 32
```

```python
# ==========================================================
#           EMPLOYEE MANAGEMENT SYSTEM
# ==========================================================
# SECTION 7 : LISTS
# ==========================================================

# ----------------------------------------------------------
# Objective:
# Learn how Python lists are used to store and manage
# multiple employee records.
#
# Topics Covered:
# 1. Create List
# 2. Append
# 3. Insert
# 4. Remove
# 5. Pop
# 6. Sort
# 7. Reverse
# 8. Slicing
# 9. Loop Through List
# ----------------------------------------------------------

print("=" * 60)
print("SECTION 7 : LISTS")
print("=" * 60)

# ==========================================================
# 1. CREATE LIST
# ==========================================================
# A list is used to store multiple values in a single variable.

employees = [
    "Rahul Sharma",
    "Priya Verma",
    "Amit Patel",
    "Sneha Joshi"
]

print("\n")
print("=" * 60)
print("1. CREATE LIST")
print("=" * 60)

print("Employee List")
print(employees)

# Expected Output:
# Employee List
# ['Rahul Sharma', 'Priya Verma', 'Amit Patel', 'Sneha Joshi']


# ==========================================================
# 2. APPEND
# ==========================================================
# append() adds a new element at the end of the list.

print("\n")
print("=" * 60)
print("2. APPEND")
print("=" * 60)

employees.append("Rohit Kumar")

print("After Appending a New Employee")
print(employees)

# Expected Output:
# ['Rahul Sharma', 'Priya Verma', 'Amit Patel',
#  'Sneha Joshi', 'Rohit Kumar']


# ==========================================================
# 3. INSERT
# ==========================================================
# insert(index, value) adds an element at a specific position.

print("\n")
print("=" * 60)
print("3. INSERT")
print("=" * 60)

employees.insert(2, "Anjali Singh")

print("After Inserting Employee at Index 2")
print(employees)

# Expected Output:
# ['Rahul Sharma', 'Priya Verma', 'Anjali Singh',
#  'Amit Patel', 'Sneha Joshi', 'Rohit Kumar']


# ==========================================================
# 4. REMOVE
# ==========================================================
# remove(value) removes the first matching element.

print("\n")
print("=" * 60)
print("4. REMOVE")
print("=" * 60)

employees.remove("Sneha Joshi")

print("After Removing Sneha Joshi")
print(employees)

# Expected Output:
# ['Rahul Sharma', 'Priya Verma',
#  'Anjali Singh', 'Amit Patel', 'Rohit Kumar']


# ==========================================================
# 5. POP
# ==========================================================
# pop() removes and returns an element.
# By default, it removes the last element.

print("\n")
print("=" * 60)
print("5. POP")
print("=" * 60)

removed_employee = employees.pop()

print(f"Removed Employee : {removed_employee}")
print("Updated Employee List")
print(employees)

# Expected Output:
# Removed Employee : Rohit Kumar
# ['Rahul Sharma', 'Priya Verma',
#  'Anjali Singh', 'Amit Patel']


# ==========================================================
# 6. SORT
# ==========================================================
# sort() arranges the list in ascending order.

print("\n")
print("=" * 60)
print("6. SORT")
print("=" * 60)

employees.sort()

print("Employees After Sorting")
print(employees)

# Expected Output:
# ['Amit Patel', 'Anjali Singh',
#  'Priya Verma', 'Rahul Sharma']


# ==========================================================
# 7. REVERSE
# ==========================================================
# reverse() reverses the current order of the list.

print("\n")
print("=" * 60)
print("7. REVERSE")
print("=" * 60)

employees.reverse()

print("Employees After Reversing")
print(employees)

# Expected Output:
# ['Rahul Sharma', 'Priya Verma',
#  'Anjali Singh', 'Amit Patel']


# ==========================================================
# 8. SLICING
# ==========================================================
# Slicing is used to access a portion of a list.

print("\n")
print("=" * 60)
print("8. SLICING")
print("=" * 60)

print("Complete Employee List")
print(employees)

print("\nFirst Two Employees")
print(employees[:2])

print("\nLast Two Employees")
print(employees[-2:])

print("\nEmployees from Index 1 to 3")
print(employees[1:4])

# Expected Output:
# First Two Employees
# ['Rahul Sharma', 'Priya Verma']
#
# Last Two Employees
# ['Anjali Singh', 'Amit Patel']
#
# Employees from Index 1 to 3
# ['Priya Verma', 'Anjali Singh', 'Amit Patel']


# ==========================================================
# 9. LOOP THROUGH LIST
# ==========================================================
# A for loop can be used to iterate through every element.

print("\n")
print("=" * 60)
print("9. LOOP THROUGH LIST")
print("=" * 60)

print("Employee Names")

for employee in employees:
    print(employee)

# Expected Output:
# Rahul Sharma
# Priya Verma
# Anjali Singh
# Amit Patel


# Loop using enumerate() to get index and value.

print("\nEmployee Number and Name")

for index, employee in enumerate(employees, start=1):
    print(f"{index}. {employee}")

# Expected Output:
# 1. Rahul Sharma
# 2. Priya Verma
# 3. Anjali Singh
# 4. Amit Patel


# ==========================================================
# Real-World Example
# ==========================================================
# Store employee salaries in a list and calculate
# the total salary expense.

print("\n")
print("=" * 60)
print("Employee Salary Report")
print("=" * 60)

employee_salaries = [
    65000,
    55000,
    70000,
    60000
]

print("Employee Salaries")
print(employee_salaries)

total_salary = sum(employee_salaries)

print(f"\nTotal Salary Expense : ₹{total_salary}")

average_salary = total_salary / len(employee_salaries)

print(f"Average Salary       : ₹{average_salary:.2f}")

# Expected Output:
# Employee Salaries
# [65000, 55000, 70000, 60000]
#
# Total Salary Expense : ₹250000
# Average Salary       : ₹62500.00
```

```python
# ==========================================================
#           EMPLOYEE MANAGEMENT SYSTEM
# ==========================================================
# SECTION 8 : TUPLES
# ==========================================================

# ----------------------------------------------------------
# Objective:
# Learn how tuples are used to store fixed (immutable)
# employee information.
#
# Topics Covered:
# 1. Create Tuple
# 2. Access Tuple Elements
# 3. count()
# 4. index()
# 5. Tuple Concatenation
# ----------------------------------------------------------

print("=" * 60)
print("SECTION 8 : TUPLES")
print("=" * 60)

# ==========================================================
# 1. CREATE TUPLE
# ==========================================================
# A tuple is an ordered collection that cannot be modified
# after it is created (immutable).

employee = (
    1001,
    "Rahul Sharma",
    "Software Development",
    "Software Engineer",
    65000
)

print("\n")
print("=" * 60)
print("1. CREATE TUPLE")
print("=" * 60)

print("Employee Record")
print(employee)

print("\nType of employee tuple:", type(employee))

# Expected Output:
# Employee Record
# (1001, 'Rahul Sharma', 'Software Development',
#  'Software Engineer', 65000)
#
# Type of employee tuple: <class 'tuple'>


# ==========================================================
# 2. ACCESS TUPLE ELEMENTS
# ==========================================================
# Tuple elements are accessed using their index.

print("\n")
print("=" * 60)
print("2. ACCESS TUPLE ELEMENTS")
print("=" * 60)

print(f"Employee ID          : {employee[0]}")
print(f"Employee Name        : {employee[1]}")
print(f"Department           : {employee[2]}")
print(f"Designation          : {employee[3]}")
print(f"Salary               : ₹{employee[4]}")

# Expected Output:
# Employee ID          : 1001
# Employee Name        : Rahul Sharma
# Department           : Software Development
# Designation          : Software Engineer
# Salary               : ₹65000


# ==========================================================
# 3. COUNT()
# ==========================================================
# count() returns the number of times a value appears
# inside the tuple.

print("\n")
print("=" * 60)
print("3. COUNT()")
print("=" * 60)

departments = (
    "HR",
    "Finance",
    "HR",
    "Testing",
    "HR",
    "Marketing"
)

print("Departments")
print(departments)

hr_count = departments.count("HR")

print(f"\nNumber of HR Departments : {hr_count}")

# Expected Output:
# Number of HR Departments : 3


# ==========================================================
# 4. INDEX()
# ==========================================================
# index() returns the position of the first occurrence
# of a specified value.

print("\n")
print("=" * 60)
print("4. INDEX()")
print("=" * 60)

position = departments.index("Testing")

print(f"'Testing' Department Found at Index : {position}")

# Expected Output:
# 'Testing' Department Found at Index : 3


# ==========================================================
# 5. TUPLE CONCATENATION
# ==========================================================
# Two tuples can be combined using the + operator.

print("\n")
print("=" * 60)
print("5. TUPLE CONCATENATION")
print("=" * 60)

employee_basic_info = (
    1001,
    "Rahul Sharma"
)

employee_job_info = (
    "Software Development",
    "Software Engineer",
    65000
)

employee_complete_record = employee_basic_info + employee_job_info

print("Basic Employee Information")
print(employee_basic_info)

print("\nJob Information")
print(employee_job_info)

print("\nComplete Employee Record")
print(employee_complete_record)

# Expected Output:
# Basic Employee Information
# (1001, 'Rahul Sharma')
#
# Job Information
# ('Software Development', 'Software Engineer', 65000)
#
# Complete Employee Record
# (1001, 'Rahul Sharma',
#  'Software Development',
#  'Software Engineer',
#  65000)


# ==========================================================
# Real-World Example
# ==========================================================
# Store multiple employee records using tuples.

print("\n")
print("=" * 60)
print("Employee Records")
print("=" * 60)

employees = [
    (1001, "Rahul Sharma", "Software Development", 65000),
    (1002, "Priya Verma", "Human Resources", 55000),
    (1003, "Amit Patel", "Finance", 70000),
    (1004, "Sneha Joshi", "Marketing", 60000)
]

for employee in employees:
    print(
        f"ID: {employee[0]} | "
        f"Name: {employee[1]} | "
        f"Department: {employee[2]} | "
        f"Salary: ₹{employee[3]}"
    )

# Expected Output:
# ID: 1001 | Name: Rahul Sharma | Department: Software Development | Salary: ₹65000
# ID: 1002 | Name: Priya Verma | Department: Human Resources | Salary: ₹55000
# ID: 1003 | Name: Amit Patel | Department: Finance | Salary: ₹70000
# ID: 1004 | Name: Sneha Joshi | Department: Marketing | Salary: ₹60000
```

```python
# ==========================================================
#           EMPLOYEE MANAGEMENT SYSTEM
# ==========================================================
# SECTION 9 : DICTIONARIES
# ==========================================================

# ----------------------------------------------------------
# Objective:
# Learn how dictionaries are used to store employee
# information using key-value pairs.
#
# Topics Covered:
# 1. Create Dictionary
# 2. Add Data
# 3. Update Data
# 4. Delete Data
# 5. Loop Through Dictionary
# 6. Nested Dictionaries
# ----------------------------------------------------------

print("=" * 60)
print("SECTION 9 : DICTIONARIES")
print("=" * 60)

# ==========================================================
# 1. CREATE DICTIONARY
# ==========================================================
# A dictionary stores data in key-value pairs.

employee = {
    "employee_id": 1001,
    "employee_name": "Rahul Sharma",
    "department": "Software Development",
    "designation": "Software Engineer",
    "salary": 65000,
    "experience": 5
}

print("\n")
print("=" * 60)
print("1. CREATE DICTIONARY")
print("=" * 60)

print("Employee Dictionary")
print(employee)

# Accessing dictionary values using keys
print("\nEmployee Details")
print(f"Employee ID   : {employee['employee_id']}")
print(f"Employee Name : {employee['employee_name']}")
print(f"Department    : {employee['department']}")
print(f"Salary        : ₹{employee['salary']}")

# Expected Output:
# Employee Dictionary
# {'employee_id':1001,
#  'employee_name':'Rahul Sharma',
#  'department':'Software Development',
#  'designation':'Software Engineer',
#  'salary':65000,
#  'experience':5}


# ==========================================================
# 2. ADD DATA
# ==========================================================
# A new key-value pair can be added to the dictionary.

print("\n")
print("=" * 60)
print("2. ADD DATA")
print("=" * 60)

employee["email"] = "rahul.sharma@company.com"
employee["phone"] = "9876543210"

print("Dictionary After Adding New Fields")
print(employee)

# Expected Output:
# email and phone are added to the dictionary.


# ==========================================================
# 3. UPDATE DATA
# ==========================================================
# Existing values can be modified.

print("\n")
print("=" * 60)
print("3. UPDATE DATA")
print("=" * 60)

employee["salary"] = 70000
employee["designation"] = "Senior Software Engineer"

print("Dictionary After Updating Data")
print(employee)

print(f"\nUpdated Salary      : ₹{employee['salary']}")
print(f"Updated Designation : {employee['designation']}")

# Expected Output:
# Salary becomes 70000
# Designation becomes Senior Software Engineer


# ==========================================================
# 4. DELETE DATA
# ==========================================================
# Data can be removed using del or pop().

print("\n")
print("=" * 60)
print("4. DELETE DATA")
print("=" * 60)

removed_phone = employee.pop("phone")

print(f"Removed Phone Number : {removed_phone}")

del employee["experience"]

print("\nDictionary After Deleting Data")
print(employee)

# Expected Output:
# Phone number removed
# Experience key deleted


# ==========================================================
# 5. LOOP THROUGH DICTIONARY
# ==========================================================
# Dictionaries can be traversed using a for loop.

print("\n")
print("=" * 60)
print("5. LOOP THROUGH DICTIONARY")
print("=" * 60)

print("Employee Information")

for key, value in employee.items():
    print(f"{key} : {value}")

# Expected Output:
# employee_id : 1001
# employee_name : Rahul Sharma
# department : Software Development
# designation : Senior Software Engineer
# salary : 70000
# email : rahul.sharma@company.com


# ==========================================================
# 6. NESTED DICTIONARIES
# ==========================================================
# A dictionary can contain other dictionaries.

print("\n")
print("=" * 60)
print("6. NESTED DICTIONARIES")
print("=" * 60)

employees = {
    1001: {
        "name": "Rahul Sharma",
        "department": "Software Development",
        "salary": 70000
    },

    1002: {
        "name": "Priya Verma",
        "department": "Human Resources",
        "salary": 55000
    },

    1003: {
        "name": "Amit Patel",
        "department": "Finance",
        "salary": 72000
    },

    1004: {
        "name": "Sneha Joshi",
        "department": "Marketing",
        "salary": 60000
    }
}

print("Complete Employee Database")
print(employees)

print("\nAccessing a Specific Employee")

print(f"Employee Name : {employees[1002]['name']}")
print(f"Department    : {employees[1002]['department']}")
print(f"Salary        : ₹{employees[1002]['salary']}")

# Expected Output:
# Employee Name : Priya Verma
# Department    : Human Resources
# Salary        : ₹55000


print("\nDisplaying All Employees")

for employee_id, details in employees.items():

    print("-" * 40)
    print(f"Employee ID : {employee_id}")
    print(f"Name        : {details['name']}")
    print(f"Department  : {details['department']}")
    print(f"Salary      : ₹{details['salary']}")

# Expected Output:
#
# ----------------------------------------
# Employee ID : 1001
# Name        : Rahul Sharma
# Department  : Software Development
# Salary      : ₹70000
#
# ----------------------------------------
# Employee ID : 1002
# Name        : Priya Verma
# Department  : Human Resources
# Salary      : ₹55000
#
# ----------------------------------------
# Employee ID : 1003
# Name        : Amit Patel
# Department  : Finance
# Salary      : ₹72000
#
# ----------------------------------------
# Employee ID : 1004
# Name        : Sneha Joshi
# Department  : Marketing
# Salary      : ₹60000


# ==========================================================
# Real-World Example
# ==========================================================
# Calculate the total monthly salary of all employees.

print("\n")
print("=" * 60)
print("Employee Salary Report")
print("=" * 60)

total_salary = 0

for details in employees.values():
    total_salary += details["salary"]

average_salary = total_salary / len(employees)

print(f"Total Employees      : {len(employees)}")
print(f"Total Monthly Salary : ₹{total_salary}")
print(f"Average Salary       : ₹{average_salary:.2f}")

# Expected Output:
# Total Employees      : 4
# Total Monthly Salary : ₹257000
# Average Salary       : ₹64250.00
```

```python
# ==========================================================
#           EMPLOYEE MANAGEMENT SYSTEM
# ==========================================================
# SECTION 10 : SETS
# ==========================================================

# ----------------------------------------------------------
# Objective:
# Learn how Python sets are used to store unique values in an
# Employee Management System.
#
# Topics Covered:
# 1. Add
# 2. Remove
# 3. Union
# 4. Intersection
# 5. Difference
#
# Note:
# - Sets store only unique values.
# - Sets are unordered, so the output order may vary.
# ----------------------------------------------------------

print("=" * 60)
print("SECTION 10 : SETS")
print("=" * 60)

# ==========================================================
# Create Sets
# ==========================================================
# Suppose employees are assigned different technical skills.

software_team_skills = {
    "Python",
    "SQL",
    "Git",
    "AWS"
}

data_team_skills = {
    "Python",
    "SQL",
    "Power BI",
    "Excel"
}

print("\nSoftware Team Skills")
print(software_team_skills)

print("\nData Team Skills")
print(data_team_skills)

# Expected Output:
# Software Team Skills
# {'Python', 'SQL', 'Git', 'AWS'}
#
# Data Team Skills
# {'Python', 'SQL', 'Power BI', 'Excel'}
#
# Note:
# Order may be different because sets are unordered.


# ==========================================================
# 1. ADD
# ==========================================================
# add() inserts a new unique element into the set.

print("\n")
print("=" * 60)
print("1. ADD")
print("=" * 60)

software_team_skills.add("Docker")

print("After Adding Docker")
print(software_team_skills)

# Adding an existing element has no effect.
software_team_skills.add("Python")

print("\nAfter Adding Python Again")
print(software_team_skills)

# Expected Output:
# Docker is added.
# Python is not duplicated.


# ==========================================================
# 2. REMOVE
# ==========================================================
# remove() deletes a specific element from the set.

print("\n")
print("=" * 60)
print("2. REMOVE")
print("=" * 60)

software_team_skills.remove("Git")

print("After Removing Git")
print(software_team_skills)

# Expected Output:
# Git is removed from the set.


# ==========================================================
# 3. UNION
# ==========================================================
# union() combines all unique elements from both sets.

print("\n")
print("=" * 60)
print("3. UNION")
print("=" * 60)

all_skills = software_team_skills.union(data_team_skills)

print("All Unique Skills")
print(all_skills)

# Expected Output:
# {'Python', 'SQL', 'AWS', 'Docker',
#  'Power BI', 'Excel'}
#
# Order may vary.


# ==========================================================
# 4. INTERSECTION
# ==========================================================
# intersection() returns the common elements
# present in both sets.

print("\n")
print("=" * 60)
print("4. INTERSECTION")
print("=" * 60)

common_skills = software_team_skills.intersection(data_team_skills)

print("Common Skills")
print(common_skills)

# Expected Output:
# {'Python', 'SQL'}
#
# Order may vary.


# ==========================================================
# 5. DIFFERENCE
# ==========================================================
# difference() returns elements that are present in
# the first set but not in the second.

print("\n")
print("=" * 60)
print("5. DIFFERENCE")
print("=" * 60)

software_only = software_team_skills.difference(data_team_skills)

print("Skills Available Only in Software Team")
print(software_only)

# Expected Output:
# {'AWS', 'Docker'}
#
# Order may vary.


# ==========================================================
# Real-World Example
# ==========================================================
# Find common and unique skills between two departments.

print("\n")
print("=" * 60)
print("Department Skill Analysis")
print("=" * 60)

print("Software Team Skills")
print(software_team_skills)

print("\nData Team Skills")
print(data_team_skills)

print("\nCommon Skills")
print(software_team_skills.intersection(data_team_skills))

print("\nAll Skills")
print(software_team_skills.union(data_team_skills))

print("\nSoftware Team Exclusive Skills")
print(software_team_skills.difference(data_team_skills))

print("\nData Team Exclusive Skills")
print(data_team_skills.difference(software_team_skills))

# Expected Output:
#
# Common Skills
# {'Python', 'SQL'}
#
# All Skills
# {'Python', 'SQL', 'AWS',
#  'Docker', 'Power BI', 'Excel'}
#
# Software Team Exclusive Skills
# {'AWS', 'Docker'}
#
# Data Team Exclusive Skills
# {'Power BI', 'Excel'}
#
# Note:
# The order of elements may differ because sets are unordered.
```

```python
# ==========================================================
#           EMPLOYEE MANAGEMENT SYSTEM
# ==========================================================
# SECTION 11 : STRING METHODS
# ==========================================================

# ----------------------------------------------------------
# Objective:
# Learn how commonly used string methods help process and
# manipulate employee information.
#
# Topics Covered:
# 1. upper()
# 2. lower()
# 3. title()
# 4. capitalize()
# 5. replace()
# 6. split()
# 7. join()
# 8. strip()
# 9. startswith()
# 10. endswith()
# 11. find()
# ----------------------------------------------------------

print("=" * 60)
print("SECTION 11 : STRING METHODS")
print("=" * 60)

# ==========================================================
# Employee Data
# ==========================================================

employee_name = "  rahul sharma  "
department = "software development"
employee_email = "rahul.sharma@company.com"

print("\nOriginal Employee Name :", repr(employee_name))
print("Department             :", department)
print("Email                  :", employee_email)

# ==========================================================
# 1. strip()
# ==========================================================
# strip() removes leading and trailing spaces.

print("\n")
print("=" * 60)
print("1. strip()")
print("=" * 60)

clean_name = employee_name.strip()

print("Original Name :", repr(employee_name))
print("Clean Name    :", repr(clean_name))

# Expected Output:
# Original Name : '  rahul sharma  '
# Clean Name    : 'rahul sharma'


# ==========================================================
# 2. upper()
# ==========================================================
# upper() converts all characters to uppercase.

print("\n")
print("=" * 60)
print("2. upper()")
print("=" * 60)

print(clean_name.upper())

# Expected Output:
# RAHUL SHARMA


# ==========================================================
# 3. lower()
# ==========================================================
# lower() converts all characters to lowercase.

print("\n")
print("=" * 60)
print("3. lower()")
print("=" * 60)

print(clean_name.lower())

# Expected Output:
# rahul sharma


# ==========================================================
# 4. title()
# ==========================================================
# title() capitalizes the first letter of each word.

print("\n")
print("=" * 60)
print("4. title()")
print("=" * 60)

formatted_name = clean_name.title()

print(formatted_name)

# Expected Output:
# Rahul Sharma


# ==========================================================
# 5. capitalize()
# ==========================================================
# capitalize() capitalizes only the first letter.

print("\n")
print("=" * 60)
print("5. capitalize()")
print("=" * 60)

print(department.capitalize())

# Expected Output:
# Software development


# ==========================================================
# 6. replace()
# ==========================================================
# replace() replaces one substring with another.

print("\n")
print("=" * 60)
print("6. replace()")
print("=" * 60)

updated_department = department.replace(
    "software",
    "product"
)

print("Original Department :", department)
print("Updated Department  :", updated_department)

# Expected Output:
# Original Department : software development
# Updated Department  : product development


# ==========================================================
# 7. split()
# ==========================================================
# split() converts a string into a list.

print("\n")
print("=" * 60)
print("7. split()")
print("=" * 60)

name_parts = clean_name.split()

print(name_parts)

# Expected Output:
# ['rahul', 'sharma']


# ==========================================================
# 8. join()
# ==========================================================
# join() combines list elements into a single string.

print("\n")
print("=" * 60)
print("8. join()")
print("=" * 60)

employee_code = "-".join(name_parts)

print(employee_code)

# Expected Output:
# rahul-sharma


# ==========================================================
# 9. startswith()
# ==========================================================
# startswith() checks whether a string begins with
# a specified value.

print("\n")
print("=" * 60)
print("9. startswith()")
print("=" * 60)

print(employee_email.startswith("rahul"))

# Expected Output:
# True


# ==========================================================
# 10. endswith()
# ==========================================================
# endswith() checks whether a string ends with
# a specified value.

print("\n")
print("=" * 60)
print("10. endswith()")
print("=" * 60)

print(employee_email.endswith(".com"))

# Expected Output:
# True


# ==========================================================
# 11. find()
# ==========================================================
# find() returns the index of the first occurrence of
# a substring.

print("\n")
print("=" * 60)
print("11. find()")
print("=" * 60)

position = employee_email.find("@")

print("Position of '@' :", position)

# Expected Output:
# Position of '@' : 13


# ==========================================================
# Real-World Example
# ==========================================================
# Generate an official employee username.

print("\n")
print("=" * 60)
print("Employee Username Generator")
print("=" * 60)

employee_name = "  Priya Verma  "

# Remove extra spaces
employee_name = employee_name.strip()

# Convert to lowercase
employee_name = employee_name.lower()

# Split into words
words = employee_name.split()

# Join using underscore
username = "_".join(words)

print(f"Employee Name : {employee_name.title()}")
print(f"Username      : {username}")
print(f"Official Email: {username}@company.com")

# Expected Output:
# Employee Name : Priya Verma
# Username      : priya_verma
# Official Email: priya_verma@company.com
```
